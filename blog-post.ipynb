### Look at how world record performances fit on this curve

While no single athlete 
would be ready for events from the 100m through the marathon, 
we can consider the best times in event as a measure of what is possible. 
World record performances are here on [Wikipedia](https://en.wikipedia.org/wiki/List_of_world_records_in_athletics#Men), 
and we convert to CSV using [this webpage](https://wikitable2csv.ggor.de/). 
We end up with data like this:

<img src="table.png" width="600" />

And on a graph, that looks like:

<img src="all-running-dedup.png" width="600" />

While that curve doesn't look special to most, 
to those familiar with [Complex Systems](vermontcomplexsystems.org) 
it has the makings of a [power law](wikipedia.org/power_law).
Work has already been done [[1,2,3]](#references) to show that the a power law model
will fit this data,
and let's take a look for ourselves.
The first thing we need to is plot this data in log-log space:

<img src="all-running-dedup-loglog.png" width="600" />

Hey, it looks like a straight line now!
We can fit a simple regression line to the data,
and that will give us the coefficients on
the power law model.
The regression in log space gives us a fit like this:

<img src="all-running-dedup-loglog-fit.png" width="600" />

We found coefficients for pace = 17.85 x distance ^ (−0.1144),
